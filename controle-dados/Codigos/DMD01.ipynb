```python
import numpy as np
from scipy.linalg import expm

# 1. Definir Sistema de 5a Ordem
n = 5
dt = 0.1
t = np.arange(0, 10, dt)
A_true = np.array([
    [0.1, -0.5, 0, 0, 0],
    [0.5, 0.1, 0, 0, 0],
    [0, 0, -0.2, 0.8, 0],
    [0, 0, -0.8, -0.2, 0],
    [0, 0, 0, 0, -0.3]
])
sys_disc = expm(A_true * dt)

# 2. Gerar Dados
x0 = np.array([1, 0.5, -1, 0, 2])
X = np.zeros((n, len(t)))
X[:, 0] = x0
for k in range(len(t) - 1):
    X[:, k+1] = sys_disc @ X[:, k]

# 3. Algoritmo DMD
X1 = X[:, :-1]
X2 = X[:, 1:]
r = 5

# Passo 1: SVD
U, S, VT = np.linalg.svd(X1, full_matrices=False)
Ur = U[:, :r]
Sr = np.diag(S[:r])
Vr = VT[:r, :].T

# Passo 2: Atilde
Atilde = Ur.T @ X2 @ Vr @ np.linalg.inv(Sr)

# Passo 3: Eigendecomposition
Lambda, W = np.linalg.eig(Atilde)

# Passo 4: Modos DMD Exatos
Phi = X2 @ Vr @ np.linalg.inv(Sr) @ W

print("Autovalores Identificados (Discretos):\n", Lambda)
```
